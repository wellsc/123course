```python
from sympy import *
a,al,b,be,c,d,r,s,t,u,v,x,y,z,w=var('a alpha b beta c d r s t u v x y z w',real=True)
from IPython.display import Markdown,Latex,display
```

```python
# rise=0.2
# run=3.0
rise=var('v_y',real=True)
run=var('v_x',real=True)
mass=15
g=9.8
ug=Matrix([0,-1])
Fg=mass*g*ug
Fg
```

```python
R=Matrix([run,rise])
R
```

```python
FR=R.dot(Fg)/R.dot(R)*R
FR
```

```python
uR=R.normalized()
uR
```

```python
Fp=Fg-FR
Fp
```

```python
simplify(Fp.dot(FR))
```

```python
Fh=Matrix([al,0])
Fh
```

```python
FhR=R.dot(Fh)/R.dot(R)*R
FhR
```

```python
al*run*uR
```

```python
-147*rise*uR
```

```python
display(FhR)
display(FR)
```

```python
eqs=[Eq(FhR[_j],-FR[_j]) for _j in range(2)]
for _j in range(len(eqs)):
    display(eqs[_j])
```

```python
al0=al.subs(solve(eqs[0])[0])
display(Markdown(r'$\alpha ='+f'{latex(al0)}'+r'$'))
```

```python
al1=al.subs(solve(eqs[1])[0])
display(Markdown(r'$\alpha ='+f'{latex(al1)}'+r'$'))
```

```python
al1.subs([(rise,0.2),(run,3.0)])
```
