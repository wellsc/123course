```python
from sympy import *
def apply(func,eqn,*args):
    return Eq(func(eqn.lhs,*args),func(eqn.rhs,*args))
x,y,z,a,b,c,th1,th2=var('x y z a b c theta_1 theta_2',real=True)
```

```python
d=60
l1=30
l2=50
leq1=Eq(l1**2+d**2-2*l1*d*cos(th1),l2**2)
leq1
```

```python
leq1=apply(lambda x:4500-x,leq1)
leq1
```

```python
leq1=apply(lambda x:x/3600,leq1)
leq1
```
