```python
from sympy import *
from IPython.display import display, Markdown, Latex
a,b,c,d=var('a b c d',positive=True)
A,B,C,alpha,beta,gamma=var('A B C alpha beta gamma',real=True)
def apply(func,eqn,*args):
    return Eq(func(eqn.lhs,*args),func(eqn.rhs,*args))
```

```python
a,b=20,30
C=pi/12
eq1=Eq(c**2,a**2+b**2-2*a*b*cos(C))
eq1
```

```python
c=simplify(solve(eq1,c)[0].subs([(a,20),(b,30)]))
c
```

```python
N(c)
```

```python
eqA=Eq(sin(A)/a,sin(C)/c)
eqA
```

```python
eqA=simplify(eqA)
eqA
```

```python
eqA=apply(lambda x:a*x,eqA)
eqA
```

```python
vA=asin(eqA.rhs)
vA
```

```python
eqB=Eq(sin(B)/b,sin(C)/c).subs(b,30)
eqB
```

```python
solve(eqB)
```

```python
eqB=apply(lambda x:30*x,eqB)
eqB=simplify(eqB)
eqB
```

```python
vB=pi-asin(eqB.rhs)
vB
```

```python
ang_sum=vA+vB+C
ang_sum
```

```python
N(ang_sum)
```
